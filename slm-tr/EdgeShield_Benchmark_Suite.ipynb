## 3. 📥 Data Ingestion & Temporal Sliding Window Normalizer
Normalizes heterogenous event schemas into standardized sliding window representation:
`[Event T-2] ... [Event T-1] ... [Target Event T_0]`
Preserves chronological cross-process and authentication context.


```python
def resolve_dataset_file(selection, custom_path=None):
    """Resolves local, cloud, or Kaggle path to the chosen dataset."""
    search_dirs = [
        'data', 'slm-tr/data',
        '/kaggle/working/slm-gpu-trainer/slm-tr/data',
        '/kaggle/input/lateral-movement-dataset-lmd-2023',
        os.path.join(repo_dir, 'data'),
        os.path.join(repo_dir, 'slm-tr/data')
    ]
    
    file_map = {
        "DARPA_OPTC": "optc_test_benchmark.csv",
        "WINDOWS_APT_DEC22": "windows_apt/22-December.csv",
        "WINDOWS_APT_NOV30": "windows_apt/30-November.csv",
        "WINDOWS_APT_DEC10": "windows_apt/10-December-P2.csv",
        "WINDOWS_APT_OCT28": "windows_apt/28-October-01-November.csv",
        "MORDOR_WINRM": "covenant_winrm.json",
        "MORDOR_WMI": "covenant_wmi.json",
        "LMD_2023_HELD_OUT": "lmd_2023_dataset.csv",
        "RANSOMWARE_STREAM": "ransomware_behavioral_traces.csv"
    }
    
    if selection == "CUSTOM":
        if os.path.exists(custom_path):
            return custom_path
        raise FileNotFoundError(f"Custom dataset path not found: {custom_path}")
        
    target_rel = file_map.get(selection)
    if not target_rel:
        raise ValueError(f"Unknown dataset selection: {selection}")
        
    for base in search_dirs:
        candidate = os.path.join(base, target_rel)
        if os.path.exists(candidate):
            return candidate
            
    # Try searching by basename recursively
    target_name = os.path.basename(target_rel)
    for root, _, files in os.walk(repo_dir):
        if target_name in files:
            return os.path.join(root, target_name)
            
    raise FileNotFoundError(f"Could not locate dataset file for '{selection}'. Searched in {search_dirs}")

data_path = resolve_dataset_file(DATASET_SELECTION, CUSTOM_DATASET_PATH)
print(f"[+] Successfully resolved dataset file: {data_path} ({os.path.getsize(data_path)/1024:.1f} KB)")
```

```python
from v2_sliding_window.data_loader import format_event_text, generate_windowed_dataframe

def load_and_preprocess_dataset(filepath, window_size=3, max_samples=None):
    """Loads CSV or JSON telemetry and generates sliding window text representations."""
    if filepath.endswith('.json') or filepath.endswith('.jsonl'):
        records = []
        with open(filepath, 'r', encoding='utf-8') as f:
            try:
                records = json.load(f)
            except Exception:
                f.seek(0)
                for line in f:
                    line = line.strip()
                    if line:
                        try:
                            records.append(json.loads(line))
                        except Exception:
                            pass
        df = pd.DataFrame(records)
    else:
        df = pd.read_csv(filepath, low_memory=False)
        
    print(f"[*] Ingested raw telemetry: {len(df):,} events. Columns: {list(df.columns[:8])}...")
    
    # Standardize column mappings across diverse schemas
    col_mappings = {
        'Image': ['image', 'ProcessName', 'process_name', 'SourceImage', '_source.data.win.eventdata.image'],
        'CommandLine': ['command_line', 'CommandLine', 'cmd', 'ProcessCommandLine', '_source.data.win.eventdata.commandLine'],
        'ParentImage': ['parent_image', 'ParentProcessName', 'ParentImage', '_source.data.win.eventdata.parentImage'],
        'ParentCommandLine': ['parent_command_line', 'ParentCommandLine', '_source.data.win.eventdata.parentCommandLine'],
        'User': ['user', 'TargetUserName', 'UserName', 'User', '_source.data.win.eventdata.user', '_source.data.win.eventdata.targetUserName'],
        'EventID': ['event_id', 'EventId', 'EventID', '_source.data.win.system.eventID'],
        'Label': ['label', 'Class', 'threat_label', 'traffic_type', 'norm_lbl']
    }
    
    for target_col, candidates in col_mappings.items():
        if target_col not in df.columns:
            for c in candidates:
                if c in df.columns:
                    df[target_col] = df[c]
                    break
                    
    # Fill defaults for missing fields
    if 'EventID' not in df.columns: df['EventID'] = 1
    if 'Image' not in df.columns: df['Image'] = 'system.exe'
    if 'CommandLine' not in df.columns: df['CommandLine'] = ''
    if 'ParentImage' not in df.columns: df['ParentImage'] = ''
    if 'ParentCommandLine' not in df.columns: df['ParentCommandLine'] = ''
    if 'User' not in df.columns: df['User'] = 'NT AUTHORITY\\SYSTEM'
    
    # Label normalization
    if 'Label' not in df.columns:
        # Default all to 1 (Attack) for pure attack emulations like Mordor
        df['Label'] = 'Attack'
        
    # Map to standardized numeric / category
    def map_label(val):
        s = str(val).lower().strip()
        if any(norm in s for norm in ['normal', 'benign', '0']):
            return 'Normal'
        elif any(eors in s for eors in ['eors', 'remote service', 't1021', 't1047']):
            return 'EoRS'
        elif any(eoht in s for eoht in ['eoht', 'hash', 'credential', 't1003', 'sam']):
            return 'EoHT'
        elif any(lotl in s for lotl in ['lotl', 't1059', 'powershell', 'wscript']):
            return 'LotL'
        else:
            return 'Attack'
            
    df['UnifiedLabel'] = df['Label'].apply(map_label)
    
    # Subsample if requested
    if max_samples and len(df) > max_samples:
        df = df.sample(n=max_samples, random_state=42).reset_index(drop=True)
        
    # Build temporal sliding windows
    print(f"[*] Constructing temporal multi-event sliding windows (K={window_size})...")
    df = generate_windowed_dataframe(df, window_size=window_size)
    print(f"[+] Prepared {len(df):,} sliding windows. Class balance:\n{df['UnifiedLabel'].value_counts().to_dict()}")
    return df

eval_df = load_and_preprocess_dataset(data_path, window_size=SLIDING_WINDOW_K, max_samples=MAX_SAMPLES)
print("\n--- Sample Formatted Sliding Window (First Event) ---")
print(eval_df['formatted_text'].iloc[0][:300] + "...")
```

## 4. 🧠 Model Engine Loader: ONNX INT8 vs. PyTorch
Loads the trained EdgeShield model. The **INT8 ONNX model** operates entirely on edge workstations via ONNX Runtime without requiring GPU dependencies, achieving 14.16 ms average latency.


```python
from transformers import AutoTokenizer
import os, sys

def resolve_model_path(backend='ONNX_INT8'):
    model_dirs = [
        'models', 'slm-tr/models',
        '/kaggle/working/slm-gpu-trainer/slm-tr/models',
        '/kaggle/input/edgeshield-slm-training/slm-tr/models',
        '/kaggle/input/edgeshield-slm-training/models',
        '/kaggle/input/edgeshield-slm-training',
        os.path.join(repo_dir, 'models'),
        os.path.join(repo_dir, 'slm-tr/models')
    ]
    if backend == 'ONNX_INT8':
        target = 'onnx_edgeshield/edgeshield_lsa_int8.onnx'
    elif backend == 'ONNX_FP32':
        target = 'onnx_edgeshield/edgeshield_lsa.onnx'
    else:
        target = 'edgeshield_lsa_deberta'
        
    for base in model_dirs:
        cand = os.path.join(base, target)
        if os.path.exists(cand):
            return cand
            
    # Auto-extract zip archive if present in /kaggle/input or /kaggle/working
    for zip_dir in ['/kaggle/input', '/kaggle/working', '.']:
        if os.path.exists(zip_dir):
            for r, _, files in os.walk(zip_dir):
                for f in files:
                    if f.endswith('.zip') and 'edgeshield' in f.lower():
                        import zipfile
                        zip_path = os.path.join(r, f)
                        extract_to = '/kaggle/working/extracted_models'
                        if not os.path.exists(extract_to):
                            print(f"[*] Extracting models archive {zip_path} -> {extract_to}...")
                            with zipfile.ZipFile(zip_path, 'r') as zf:
                                zf.extractall(extract_to)
                        for er, _, efiles in os.walk(extract_to):
                            if os.path.basename(target) in efiles:
                                return os.path.join(er, os.path.basename(target))

    # Comprehensive recursive fallback search
    search_roots = [repo_dir]
    if os.path.exists('/kaggle/input'): search_roots.append('/kaggle/input')
    if os.path.exists('/kaggle/working'): search_roots.append('/kaggle/working')
    if os.path.exists('.'): search_roots.append('.')
    
    base_name = os.path.basename(target)
    for root_dir in search_roots:
        for root, _, files in os.walk(root_dir):
            if base_name in files:
                return os.path.join(root, base_name)
                
    raise FileNotFoundError(f"Could not locate model artifact for {backend}. Searched in {model_dirs} and {search_roots}")

model_target = resolve_model_path(BACKEND)
try:
    tokenizer_path = resolve_model_path('PYTORCH')
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_path)
    print(f"[+] Loaded Tokenizer from: {tokenizer_path}")
except Exception as e:
    print(f"[*] Local tokenizer not found ({e}). Loading DeBERTa-v3 tokenizer from HuggingFace hub...")
    tokenizer = AutoTokenizer.from_pretrained('microsoft/deberta-v3-xsmall')
    print(f"[+] Loaded Tokenizer from: microsoft/deberta-v3-xsmall")

print(f"[+] Loaded Model Target: {model_target}")

# Instantiate Engine
if 'ONNX' in BACKEND:
    import onnxruntime as ort
    sess_options = ort.SessionOptions()
    sess_options.graph_optimization_level = ort.GraphOptimizationLevel.ORT_ENABLE_ALL
    session = ort.InferenceSession(model_target, sess_options, providers=['CPUExecutionProvider'])
    print(f"[+] ONNX Runtime Session active on: {session.get_providers()}")
else:
    import torch
    from transformers import AutoModelForSequenceClassification
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    pt_model = AutoModelForSequenceClassification.from_pretrained(model_target)
    pt_model.to(device)
    pt_model.eval()
    print(f"[+] PyTorch Model active on: {device}")
```

## 5. ⚡ Live Empirical Inference Execution & Latency Profiling
Executes live inference over the dataset, capturing prediction logits, threat probability, classification labels, and millisecond execution latencies.


```python
texts = eval_df['formatted_text'].tolist()
latencies_ms = []
predictions = []
threat_probs = []

print(f"[*] Starting live empirical inference on {len(texts):,} windows (Batch Size: {BATCH_SIZE})...")
t_start = time.perf_counter()

for i in range(0, len(texts), BATCH_SIZE):
    batch_texts = texts[i : i + BATCH_SIZE]
    inputs = tokenizer(
        batch_texts,
        padding="max_length" if 'ONNX' in BACKEND else True,
        truncation=True,
        max_length=128 if 'ONNX' in BACKEND else 256,
        return_tensors="np" if 'ONNX' in BACKEND else "pt"
    )
    
    t0 = time.perf_counter()
    if 'ONNX' in BACKEND:
        ort_inputs = {
            "input_ids": inputs["input_ids"].astype(np.int64),
            "attention_mask": inputs["attention_mask"].astype(np.int64)
        }
        logits = session.run(None, ort_inputs)[0]
    else:
        with torch.no_grad():
            inp_dev = {k: v.to(device) for k, v in inputs.items()}
            logits = pt_model(**inp_dev).logits.cpu().numpy()
            
    t1 = time.perf_counter()
    batch_time_ms = (t1 - t0) * 1000.0
    latencies_ms.extend([batch_time_ms / len(batch_texts)] * len(batch_texts))
    
    # Softmax probabilities
    exp_logits = np.exp(logits - np.max(logits, axis=-1, keepdims=True))
    probs = exp_logits / np.sum(exp_logits, axis=-1, keepdims=True)
    
    preds = np.argmax(logits, axis=-1)
    predictions.extend(preds)
    
    # Threat probability = 1.0 - Normal_prob (index 0 is Normal)
    threat_probs.extend(1.0 - probs[:, 0])

total_time_s = time.perf_counter() - t_start
fps = len(texts) / total_time_s

print(f"[+] Benchmark Complete!")
print(f"    Total Time: {total_time_s:.2f} s | Throughput: {fps:.1f} events/sec")
print(f"    Average Latency: {np.mean(latencies_ms):.2f} ms/event | P95: {np.percentile(latencies_ms, 95):.2f} ms")
```

## 6. 📊 Evaluation Metrics, Attribution & Confusion Matrix
Calculates empirical performance metrics based on the selected mode:
- **Level 1 (Binary Interception)**: Normal vs. Malicious Threat / Lateral Movement
- **Level 2 (Granular Attribution)**: Per-technique MITRE classification


```python
from sklearn.metrics import classification_report, accuracy_score, precision_recall_fscore_support, confusion_matrix, roc_auc_score
import matplotlib.pyplot as plt
import seaborn as sns

eval_df['predicted_class_id'] = predictions
eval_df['threat_probability'] = threat_probs

id2label = {0: 'Normal', 1: 'EoRS', 2: 'EoHT'}
eval_df['predicted_label'] = [id2label.get(p, 'Attack') for p in predictions]

if EVALUATION_MODE == "HIERARCHICAL_LEVEL_1":
    # Binary: Normal (0) vs Attack (1)
    y_true_binary = (eval_df['UnifiedLabel'] != 'Normal').astype(int).tolist()
    y_pred_binary = (eval_df['predicted_class_id'] != 0).astype(int).tolist()
    
    acc = accuracy_score(y_true_binary, y_pred_binary)
    prec, rec, f1, _ = precision_recall_fscore_support(y_true_binary, y_pred_binary, average='binary', zero_division=0)
    cm = confusion_matrix(y_true_binary, y_pred_binary)
    tn, fp, fn, tp = cm.ravel()
    fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0
    
    try:
        auc = roc_auc_score(y_true_binary, threat_probs)
    except Exception:
        auc = 0.5
        
    print("=" * 70)
    print(f"🛡️ LEVEL 1: BINARY THREAT INTERCEPTION BENCHMARK ({DATASET_SELECTION})")
    print("=" * 70)
    print(f"Accuracy:                {acc * 100:.2f}%")
    print(f"Threat Recall (Sensitivity): {rec * 100:.2f}%  ({tp}/{tp+fn} attacks intercepted)")
    print(f"Normal Specificity:      {(1.0 - fpr) * 100:.2f}%  ({tn}/{tn+fp} normal events isolated)")
    print(f"Precision:               {prec * 100:.2f}%")
    print(f"Macro F1-Score:          {f1 * 100:.2f}%")
    print(f"False Positive Rate (FPR): {fpr * 100:.2f}%")
    print(f"ROC AUC Score:           {auc:.4f}")
    print(f"Mean Latency:            {np.mean(latencies_ms):.2f} ms/event (P95: {np.percentile(latencies_ms, 95):.2f} ms)")
    print("=" * 70)
    
    # Plot Confusion Matrix
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Normal', 'Threat Alert'], yticklabels=['Normal', 'Actual Threat'])
    plt.title(f"Level 1 Threat Interception Confusion Matrix\nDataset: {DATASET_SELECTION}")
    plt.ylabel('True Class')
    plt.xlabel('Predicted Class')
    plt.tight_layout()
    plt.show()

else:
    # Granular Multi-Class Attribution
    target_names = ['Normal', 'EoRS', 'EoHT']
    y_true_mc = [0 if l == 'Normal' else (1 if l == 'EoRS' else 2) for l in eval_df['UnifiedLabel']]
    y_pred_mc = predictions
    
    acc = accuracy_score(y_true_mc, y_pred_mc)
    cm = confusion_matrix(y_true_mc, y_pred_mc)
    
    print("=" * 70)
    print(f"🔍 LEVEL 2: MITRE ATT&CK ATTRIBUTION BENCHMARK ({DATASET_SELECTION})")
    print("=" * 70)
    print(f"Overall Accuracy: {acc * 100:.2f}%")
    print("\nDetailed Classification Report:")
    print(classification_report(y_true_mc, y_pred_mc, target_names=target_names, digits=4, zero_division=0))
    print("=" * 70)
    
    # Plot Multi-class Confusion Matrix
    plt.figure(figsize=(7, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Reds', xticklabels=target_names, yticklabels=target_names)
    plt.title(f"Level 2 Technique Attribution Confusion Matrix\nDataset: {DATASET_SELECTION}")
    plt.ylabel('True MITRE Technique')
    plt.xlabel('Attributed Technique')
    plt.tight_layout()
    plt.show()
```

## 10. ⚡ Tiered Selective Gating & Continuous Streaming Replay (EXP-RDC4-005)
### High-Throughput Triage Engine for Enterprise Endpoint Deployment

**Problem:** In continuous enterprise endpoint telemetry, 85%–95% of event streams consist of routine background OS maintenance (e.g., `SearchIndexer.exe`, routine `svchost.exe`, `RuntimeBroker.exe`). Evaluating every benign event through deep Transformer multi-head self-attention (~13.22 ms) caps throughput at ~75 EPS, creating CPU saturation during high-volume bursts.

**Solution:** The **Tiered Selective Gating Engine** introduces a two-tier triage pipeline:
1. **Tier 1 (Fast-Path Triage, < 0.05 ms / < 50 µs):**
   - Instant allowlist matching for verified background OS processes without anomalous command lines.
   - Shannon entropy analysis ($\mathcal{H} > 4.2$) to catch obfuscated/encoded payloads (Base64, XOR).
   - Fast regex scanning for critical attack indicators (`vssadmin`, `mimikatz`, `bypass`, `powershell -enc`, `certutil`, `rundll32`).
2. **Tier 2 (Deep SLM Verification via ONNX INT8, ~13 ms):**
   - Any event matching an attack indicator, displaying abnormal entropy, or originating from unverified binaries is forwarded to the quantized SLM for full contextual sequence evaluation.

This experiment replays continuous telemetry streams to empirically evaluate throughput speedup, effective latency per event, and zero-false-negative attack retention.

```python
# 10.1 Tiered Selective Gating Engine Implementation & Streaming Replay
import time
import math
import re
from typing import Dict, List, Any, Tuple, Optional
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

try:
    from edgeshield.selective_gating import TieredSelectiveGatingEngine, DEFAULT_BENIGN_PROCESSES, CRITICAL_ATTACK_INDICATORS
    print("[+] Successfully imported TieredSelectiveGatingEngine from edgeshield.selective_gating")
except ImportError:
    print("[*] Using embedded TieredSelectiveGatingEngine definition...")
    DEFAULT_BENIGN_PROCESSES = {
        'searchprotocolhost.exe', 'searchindexer.exe', 'svchost.exe', 'smartscreen.exe',
        'taskhostw.exe', 'wsqmcons.exe', 'conhost.exe', 'backgroundtaskhost.exe',
        'runtimebroker.exe', 'comppkgsrv.exe', 'dwm.exe', 'explorer.exe',
        'sihost.exe', 'fontdrvhost.exe', 'ctfmon.exe'
    }
    CRITICAL_ATTACK_INDICATORS = [
        'bypass', 'iex', 'download', 'invoke', 'powershell -enc', 'powershell -ep',
        'wmic', 'psexec', 'mimikatz', 'reg save', 'net use', 'vssadmin', 'bcdedit',
        'whoami', 'certutil', 'cscript', 'wscript', 'rundll32', 'mshta', 'regsvr32',
        'nltest', 'secretsdump', 'lsass', 'sam.save', 'system.save', 'inhibit',
        'shadowcopy', 'delete shadows', 'wevtutil', 'sc stop', 'net stop'
    ]
    class TieredSelectiveGatingEngine:
        def __init__(self, benign_allowlist=None, entropy_threshold=4.2):
            self.allowlist = benign_allowlist or DEFAULT_BENIGN_PROCESSES
            self.entropy_threshold = entropy_threshold
            self._compiled_indicators = [re.compile(re.escape(k), re.IGNORECASE) for k in CRITICAL_ATTACK_INDICATORS]

        @staticmethod
        def calculate_shannon_entropy(text: str) -> float:
            if not text or len(text) < 8: return 0.0
            prob = [float(text.count(c)) / len(text) for c in set(text)]
            return -sum(p * math.log2(p) for p in prob)

        def evaluate_event(self, record: Dict[str, Any]) -> Tuple[bool, float, str]:
            t0 = time.perf_counter()
            cmd = str(record.get('CommandLine', record.get('command_line', record.get('cmd', '')))).strip()
            image = str(record.get('Image', record.get('image', record.get('process_name', '')))).strip().lower()
            image_clean = image.replace('\\', '/').split('/')[-1]

            cmd_lower = cmd.lower()
            for pattern in self._compiled_indicators:
                if pattern.search(cmd_lower):
                    return False, (time.perf_counter() - t0) * 1e6, "CRITICAL_KEYWORD_MATCH"

            if len(cmd) > 30:
                entropy = self.calculate_shannon_entropy(cmd)
                if entropy > self.entropy_threshold:
                    return False, (time.perf_counter() - t0) * 1e6, f"HIGH_ENTROPY_OBFUSCATION ({entropy:.2f})"

            if image_clean in self.allowlist:
                return True, (time.perf_counter() - t0) * 1e6, f"ALLOWLIST_BENIGN_PROCESS ({image_clean})"

            user = str(record.get('User', record.get('user', ''))).lower()
            if (not cmd or cmd in ['nan', 'none']) and ('system' in user or 'service' in user):
                return True, (time.perf_counter() - t0) * 1e6, "BENIGN_EMPTY_SYSTEM_CALL"

            return False, (time.perf_counter() - t0) * 1e6, "UNRESOLVED_SEQUENCE_FORWARD"

        def replay_stream(self, events: List[Dict[str, Any]], slm_inference_fn=None, slm_simulated_latency_ms=13.22):
            total_events = len(events)
            fast_path_dismissed = 0
            deep_slm_forwarded = 0
            gate_latencies_us = []
            slm_latencies_ms = []
            attack_total = 0
            attack_intercepted = 0
            attack_false_negatives = 0

            t_stream_start = time.perf_counter()
            for event in events:
                lbl = str(event.get('Label', event.get('UnifiedLabel', event.get('norm_lbl', 'Normal'))))
                is_attack = lbl not in ['Normal', 'normal', '0', 0]
                if is_attack: attack_total += 1

                is_dismissed, elapsed_us, reason = self.evaluate_event(event)
                gate_latencies_us.append(elapsed_us)

                if is_dismissed:
                    fast_path_dismissed += 1
                    if is_attack: attack_false_negatives += 1
                else:
                    deep_slm_forwarded += 1
                    if is_attack: attack_intercepted += 1
                    if slm_inference_fn:
                        t_slm0 = time.perf_counter()
                        _ = slm_inference_fn(event)
                        slm_latencies_ms.append((time.perf_counter() - t_slm0) * 1000.0)
                    else:
                        slm_latencies_ms.append(slm_simulated_latency_ms)

            gating_ratio = (fast_path_dismissed / total_events) * 100.0 if total_events > 0 else 0.0
            mean_fast_ms = np.mean(gate_latencies_us) / 1000.0 if gate_latencies_us else 0.0
            mean_slm_ms = np.mean(slm_latencies_ms) if slm_latencies_ms else slm_simulated_latency_ms
            effective_latency_ms = (mean_fast_ms * (gating_ratio / 100.0)) + (mean_slm_ms * ((100.0 - gating_ratio) / 100.0))
            sustained_eps = 1000.0 / effective_latency_ms if effective_latency_ms > 0 else 0.0
            ungated_eps = 1000.0 / mean_slm_ms if mean_slm_ms > 0 else 0.0
            speedup = sustained_eps / ungated_eps if ungated_eps > 0 else 1.0

            return {
                "total_events": total_events,
                "fast_path_dismissed": fast_path_dismissed,
                "deep_slm_forwarded": deep_slm_forwarded,
                "gating_ratio_percent": round(gating_ratio, 2),
                "attacks_evaluated": attack_total,
                "attacks_routed_to_slm": attack_intercepted,
                "attacks_missed_by_gate": attack_false_negatives,
                "attack_retention_rate": round((attack_intercepted / attack_total) * 100.0, 2) if attack_total > 0 else 100.0,
                "fast_path_avg_us": round(float(np.mean(gate_latencies_us)), 2),
                "deep_slm_avg_ms": round(float(mean_slm_ms), 2),
                "effective_latency_per_event_ms": round(float(effective_latency_ms), 3),
                "sustained_throughput_eps": round(float(sustained_eps), 1),
                "ungated_throughput_eps": round(float(ungated_eps), 1),
                "throughput_speedup": round(float(speedup), 2),
                "cpu_savings_percent": round((1.0 - (1.0 / speedup)) * 100.0, 2)
            }

# Instantiate Gating Engine
gating_engine = TieredSelectiveGatingEngine()

# Live SLM Inference Callback
def run_live_slm_event(event):
    text = event.get('formatted_text', '')
    if not text:
        text = f"EventID: {event.get('EventID', 1)} | Image: {event.get('Image', '')} | CommandLine: {event.get('CommandLine', '')}"
    if 'ONNX' in BACKEND and 'session' in globals():
        inputs = tokenizer(text, max_length=128, padding='max_length', truncation=True, return_tensors='np')
        ort_inputs = {
            'input_ids': inputs['input_ids'].astype(np.int64),
            'attention_mask': inputs['attention_mask'].astype(np.int64)
        }
        if 'token_type_ids' in [x.name for x in session.get_inputs()]:
            ort_inputs['token_type_ids'] = inputs['token_type_ids'].astype(np.int64)
        return session.run(None, ort_inputs)
    elif 'pt_model' in globals():
        inputs = tokenizer(text, max_length=128, padding='max_length', truncation=True, return_tensors='pt').to(device)
        with torch.no_grad():
            return pt_model(**inputs)
    else:
        return None

# Prepare telemetry stream from loaded dataset
stream_records = eval_df.to_dict('records')
print(f"[*] Starting Continuous Telemetry Stream Replay: {len(stream_records):,} events...")

replay_metrics = gating_engine.replay_stream(
    events=stream_records,
    slm_inference_fn=run_live_slm_event,
    slm_simulated_latency_ms=float(np.mean(latencies_ms)) if 'latencies_ms' in globals() and len(latencies_ms) > 0 else 13.22
)

print("\n" + "="*80)
print(" ⚡ TIERED SELECTIVE GATING EMPIRICAL METRICS (EXP-RDC4-005)")
print("="*80)
for k, v in replay_metrics.items():
    print(f"  {k:35s}: {v}")
print("="*80)
```
